````python
class parent:
  PatternAttr=100
  def __init__(self):
    print("instituation parent")
  def parent_method(self):
    print("parent method")
  def setAttr(self,attr):
    self.PatternAttr=attr
  def getAtrr(self):
    print("set of parent attribute",self.PatternAttr)
class child(parent):
  def __init__(self):
    print("instituation childern")
  def childMethod(self):
       print('child method')
  def getAttr(self):
        print(super().getAtrr)
        print("child says hello!")
b= parent()
b.PatternAttr
c = child() # instantiate child
c.childMethod() # call child method
c.parent_method() # call parent method
c.setAttr(200) # set attribute (parent method)
c.getAttr() # call attribute (parent method)
print(c.PatternAttr)
c.childMethod() # call child method


````

Abstract

````python
from abc import ABC, abstractmethod
class Animal(ABC):
  @abstractmethod
  def eats_grass(self):
    pass
  @abstractmethod
  def is_animal(self):
    return True
class Cat(Animal):
   def __init__(self):
        print("miew~~") # overriding abstract method
   def eats_grass(self):
       return False
   def is_animal(self):
       pass

miew = Cat()
print(miew.is_animal())
print(miew.eats_grass())

````

````python
from abc import ABC, abstractmethod
class Animal(ABC):
  @abstractmethod
  def eats_grass(self):
    pass

  def is_animal(self):
    return True
class Cat(Animal):
   def __init__(self):
        print("miew~~") # overriding abstract method
   def eats_grass(self):
       return False


miew = Cat()
print(miew.is_animal())
print(miew.eats_grass())
````
