# lec7:sorting algorithms

1-buble algorithm:

```python
def bubble_sort(arr):
  n=len(arr)
  for i in range(n):
    for j in range(0,n-i-1):
      if arr[j]>arr[j+1]:
        arr[j],arr[j+1]=arr[j+1],arr[j]
arr=[20,5,3,8]
bubble_sort(arr)
print("ordered array:")
for i in range(len(arr)):
  print(arr[i])
```

2- selection sort

```python
def selection_sort(arr,size):
  for s in range (size):
    min_index=s
    for i in range(s+1,size):
      if arr[i]<arr[min_index]:
        min_index=i
        arr[s],arr[min_index]=arr[min_index],arr[s]
arr=[30,6,4,2]
size=len(arr)
selection_sort(arr,size)
print(arr)
for i in range(len(arr)):
  print(arr[i])
```

3- insertion sort

```python
def insertion_sort(list1):
  #outer loop of traverse list
  for i in range(1,len(list1)):
    a=list1[i]
    #sorted part one greater than other
    j=i-1
    while j>=0 and a<list1[j]:
      list1[j+1]=list1[j]
      j -=1
    list1[j+1]=a
  return list1
list1=[7,2,1,6]
print("the unsorted list is:",list1)
print("the new sorted list is",insertion_sort(list1))
```

2-binary classification

```python
def binary_search(arr,target,low,high):
  if low<= high:
    mid = (low+high)//2

    if arr[mid]== target:
      return mid
    elif arr[mid]<target:
      return binary_search(arr,target,mid+1,high)
    else:
      return binary_search(arr,target,low,mid-1)
  else:
    return -1
arr=[5,2,10,3]
target=3
result=binary_search(sorted(arr),target,0,len(arr)-1)
print(result)
```
